```python
# =====================NOTRANSFER-LEARNING CNN-SNN (LIF NEURONS )DATASET-1=============================
# 1. MOUNT GOOGLE DRIVE
# ===============================================================
from google.colab import drive
drive.mount('/content/drive')

# ===============================================================
# 2. INSTALL DEPENDENCIES
# ===============================================================
!pip install spikingjelly seaborn scikit-learn torchvision --quiet

# ===============================================================
# 3. DATASET PREPARATION (70/15/15)
# ===============================================================
import os
from torchvision import transforms
from torch.utils.data import DataLoader, Dataset, random_split
from PIL import Image
import torch
import torch.nn as nn

root_dir = '/content/drive/MyDrive/Dataset#1A/Dataset#1A/here/Dataset/datasetFull'

transform = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.ToTensor(),
    transforms.Normalize([0.485, 0.456, 0.406],
                         [0.229, 0.224, 0.225])
])

class ForestFireDataset(Dataset):
    def __init__(self, root_dir, transform=None):
        self.image_paths = []
        self.labels = []
        self.transform = transform

        for label_name in ['nofire', 'fire']:
            class_dir = os.path.join(root_dir, label_name)
            label = 1 if label_name == 'fire' else 0

            if not os.path.exists(class_dir):
                continue

            for root, _, files in os.walk(class_dir):
                for file_name in files:
                    if file_name.lower().endswith(('.jpg', '.jpeg', '.png')):
                        self.image_paths.append(os.path.join(root, file_name))
                        self.labels.append(label)

    def __len__(self):
        return len(self.image_paths)

    def __getitem__(self, idx):
        image = Image.open(self.image_paths[idx]).convert("RGB")
        if self.transform:
            image = self.transform(image)
        return image, self.labels[idx]

full_dataset = ForestFireDataset(root_dir, transform)

print("Total images found:", len(full_dataset))
assert len(full_dataset) > 0, "Dataset is empty! Check path or folder names."

dataset_size = len(full_dataset)
train_size = int(0.70 * dataset_size)
val_size   = int(0.15 * dataset_size)
test_size  = dataset_size - train_size - val_size

torch.manual_seed(42)
train_dataset, val_dataset, test_dataset = random_split(
    full_dataset, [train_size, val_size, test_size]
)

train_loader = DataLoader(train_dataset, batch_size=10, shuffle=True)
val_loader   = DataLoader(val_dataset, batch_size=10, shuffle=False)
test_loader  = DataLoader(test_dataset, batch_size=10, shuffle=False)

print(f"Dataset Split → Train: {len(train_dataset)}, Val: {len(val_dataset)}, Test: {len(test_dataset)}")

# ===============================================================
# 4. MODEL (CNN + 4-LAYER SNN WITH LIF)
# ===============================================================
from spikingjelly.clock_driven import neuron, surrogate, encoding, functional

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

# CNN Branch 1
class CNN_Branch1(nn.Module):
    def __init__(self):
        super().__init__()
        self.features = nn.Sequential(
            nn.Conv2d(3, 32, 3, padding=1),
            nn.BatchNorm2d(32),
            nn.ReLU(),
            nn.MaxPool2d(2),

            nn.Conv2d(32, 64, 3, padding=1),
            nn.BatchNorm2d(64),
            nn.ReLU(),
            nn.MaxPool2d(2),

            nn.Conv2d(64, 128, 3, padding=1),
            nn.BatchNorm2d(128),
            nn.ReLU(),
            nn.AdaptiveAvgPool2d((1, 1))
        )

    def forward(self, x):
        return torch.flatten(self.features(x), 1)

# CNN Branch 2
class CNN_Branch2(nn.Module):
    def __init__(self):
        super().__init__()
        self.features = nn.Sequential(
            nn.Conv2d(3, 32, 3, padding=1),
            nn.ReLU(),
            nn.MaxPool2d(2),

            nn.Conv2d(32, 64, 3, padding=1),
            nn.ReLU(),
            nn.MaxPool2d(2),

            nn.Conv2d(64, 128, 3, padding=1),
            nn.ReLU(),
            nn.MaxPool2d(2),

            nn.Conv2d(128, 256, 3, padding=1),
            nn.ReLU(),
            nn.AdaptiveAvgPool2d((1, 1))
        )

    def forward(self, x):
        return torch.flatten(self.features(x), 1)

# FULL MODEL
class CNN_SNN_Model(nn.Module):
    def __init__(self):
        super().__init__()

        self.branch1 = CNN_Branch1()
        self.branch2 = CNN_Branch2()
        self.encoder = encoding.PoissonEncoder()

        fused_dim = 128 + 256

        # 4 SNN layers
        self.fc1 = nn.Linear(fused_dim, 512)
        self.lif1 = neuron.LIFNode(surrogate_function=surrogate.Sigmoid())

        self.fc2 = nn.Linear(512, 256)
        self.lif2 = neuron.LIFNode(surrogate_function=surrogate.Sigmoid())

        self.fc3 = nn.Linear(256, 128)
        self.lif3 = neuron.LIFNode(surrogate_function=surrogate.Sigmoid())

        self.fc4 = nn.Linear(128, 2)
        self.lif4 = neuron.LIFNode(surrogate_function=surrogate.Sigmoid())

    def forward(self, x, time_window=10):
        spk_out = 0

        for t in range(time_window):
            f1 = self.branch1(x)
            f2 = self.branch2(x)

            fused = torch.cat((f1, f2), dim=1)
            encoded = self.encoder(fused)

            out = self.lif1(self.fc1(encoded))
            out = self.lif2(self.fc2(out))
            out = self.lif3(self.fc3(out))
            out = self.lif4(self.fc4(out))

            spk_out += out

        return spk_out / time_window

model = CNN_SNN_Model().to(device)

criterion = nn.CrossEntropyLoss()
optimizer = torch.optim.Adam(model.parameters(), lr=5e-4)

# ===============================================================
# 5. TRAINING
# ===============================================================
import matplotlib.pyplot as plt

train_losses, val_losses = [], []
train_accs, val_accs = [], []

def validate():
    model.eval()
    total, correct, loss_sum = 0, 0, 0

    with torch.no_grad():
        for data, target in val_loader:
            data, target = data.to(device), target.to(device)
            out = model(data)

            loss_sum += criterion(out, target).item()
            preds = torch.argmax(out, dim=1)

            correct += (preds == target).sum().item()
            total += target.size(0)

            functional.reset_net(model)

    return loss_sum / len(val_loader), 100 * correct / total

def train(epoch):
    model.train()
    total, correct, loss_sum = 0, 0, 0

    for batch_idx, (data, target) in enumerate(train_loader):
        data, target = data.to(device), target.to(device)

        optimizer.zero_grad()
        out = model(data)
        loss = criterion(out, target)
        loss.backward()

        torch.nn.utils.clip_grad_norm_(model.parameters(), 5.0)

        optimizer.step()
        functional.reset_net(model)

        loss_sum += loss.item()
        preds = torch.argmax(out, dim=1)

        correct += (preds == target).sum().item()
        total += target.size(0)

        if batch_idx % 10 == 0:
            print(f"Epoch {epoch} | Batch {batch_idx} | Loss: {loss.item():.4f}")

    train_losses.append(loss_sum / len(train_loader))
    train_accs.append(100 * correct / total)

    val_loss, val_acc = validate()
    val_losses.append(val_loss)
    val_accs.append(val_acc)

    print(f"\nEpoch {epoch} → Train Acc={train_accs[-1]:.2f}% | Val Acc={val_acc:.2f}%\n")

for epoch in range(1, 51):
    train(epoch)

# ===============================================================
# 6. EVALUATION
# ===============================================================
from sklearn.metrics import classification_report, confusion_matrix, accuracy_score
import seaborn as sns

def evaluate(loader):
    model.eval()
    y_true, y_pred = [], []

    with torch.no_grad():
        for data, labels in loader:
            data, labels = data.to(device), labels.to(device)
            outputs = model(data)

            preds = torch.argmax(outputs, dim=1)
            y_true.extend(labels.cpu().numpy())
            y_pred.extend(preds.cpu().numpy())

            functional.reset_net(model)

    return y_true, y_pred

y_true, y_pred = evaluate(test_loader)

print("Test Accuracy:", accuracy_score(y_true, y_pred))
print(classification_report(y_true, y_pred, target_names=["Not Fire", "Fire"]))

cm = confusion_matrix(y_true, y_pred)
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues',
            xticklabels=["Not Fire", "Fire"],
            yticklabels=["Not Fire", "Fire"])
plt.title("Confusion Matrix")
plt.show()

# ===============================================================
# 9. TRAINING CURVES
# ===============================================================
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

axes[0].plot(train_losses, marker='o', label='Train Loss')
axes[0].plot(val_losses,   marker='o', label='Val Loss')
axes[0].set_title('Loss vs Epoch'); axes[0].set_xlabel('Epoch')
axes[0].set_ylabel('Loss'); axes[0].grid(True); axes[0].legend()

axes[1].plot(train_accs, marker='o', label='Train Acc')
axes[1].plot(val_accs,   marker='o', label='Val Acc')
axes[1].set_title('Accuracy vs Epoch'); axes[1].set_xlabel('Epoch')
axes[1].set_ylabel('Accuracy (%)'); axes[1].grid(True); axes[1].legend()

plt.tight_layout()
plt.show()
```

```python
# =======================PROPOSED MODEL (CNN-SNN, LIF NEURON) WITH DATASET-1 — TRANSFER LEARNING=====================
# 0. MOUNT DRIVE + INSTALL
# ===============================================================
from google.colab import drive
drive.mount('/content/drive')

!pip install torch torchvision spikingjelly seaborn scikit-learn --quiet


# ===============================================================
# 1. IMPORTS
# ===============================================================
import os
import time
import copy

import torch
import torch.nn as nn
import torch.nn.functional as F
import torchvision.models as models
import torchvision.transforms as transforms
import matplotlib.pyplot as plt
import seaborn as sns

from torch.utils.data import DataLoader, Dataset, random_split, Subset
from PIL import Image
from sklearn.metrics import (confusion_matrix, classification_report,
                             accuracy_score)
from spikingjelly.clock_driven import encoding, functional


# ===============================================================
# 2. CONFIGURATION
# ===============================================================
ROOT_DIR     = '/content/drive/MyDrive/Dataset#1A/Dataset#1A/here/Dataset/datasetFull'
BATCH_SIZE   = 10
TIME_WINDOW  = 10          # kept at 10 — CNN is outside loop so cost is fixed
NUM_EPOCHS   = 50
LR           = 1e-4
SPIKE_REG    = 0.001       # spike-rate regularisation weight
TARGET_RATE  = 0.10        # target mean firing rate (10%)
PATIENCE     = 8           # early stopping patience
DEVICE       = torch.device("cuda" if torch.cuda.is_available() else "cpu")

print(f"Device     : {DEVICE}")
print(f"Time window: T={TIME_WINDOW}")
print(f"Spike reg  : λ={SPIKE_REG}, target rate={TARGET_RATE}")


# ===============================================================
# 3. DATASET
# ===============================================================
transform_train = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.RandomHorizontalFlip(),
    transforms.ColorJitter(brightness=0.2, contrast=0.2),
    transforms.ToTensor(),
    transforms.Normalize([0.485, 0.456, 0.406],
                         [0.229, 0.224, 0.225])
])

transform_eval = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.ToTensor(),
    transforms.Normalize([0.485, 0.456, 0.406],
                         [0.229, 0.224, 0.225])
])


class ForestFireDataset(Dataset):
    def __init__(self, root_dir, transform=None):
        self.image_paths = []
        self.labels      = []
        self.transform   = transform

        for label_name in ['nofire', 'fire']:
            class_dir = os.path.join(root_dir, label_name)
            label     = 1 if label_name == 'fire' else 0
            if not os.path.isdir(class_dir):
                continue
            for fname in os.listdir(class_dir):
                if fname.lower().endswith(('.jpg', '.jpeg', '.png')):
                    self.image_paths.append(os.path.join(class_dir, fname))
                    self.labels.append(label)

    def __len__(self):
        return len(self.image_paths)

    def __getitem__(self, idx):
        img = Image.open(self.image_paths[idx]).convert("RGB")
        if self.transform:
            img = self.transform(img)
        return img, self.labels[idx]


# Build index split once, apply different transforms per split
base_dataset = ForestFireDataset(ROOT_DIR)
n            = len(base_dataset)
train_size   = int(0.70 * n)
val_size     = int(0.15 * n)
test_size    = n - train_size - val_size

torch.manual_seed(42)
indices      = torch.randperm(n).tolist()
train_idx    = indices[:train_size]
val_idx      = indices[train_size:train_size + val_size]
test_idx     = indices[train_size + val_size:]

train_dataset = Subset(ForestFireDataset(ROOT_DIR, transform_train), train_idx)
val_dataset   = Subset(ForestFireDataset(ROOT_DIR, transform_eval),  val_idx)
test_dataset  = Subset(ForestFireDataset(ROOT_DIR, transform_eval),  test_idx)

train_loader = DataLoader(train_dataset, batch_size=BATCH_SIZE,
                          shuffle=True,  num_workers=2, pin_memory=True)
val_loader   = DataLoader(val_dataset,   batch_size=BATCH_SIZE,
                          shuffle=False, num_workers=2, pin_memory=True)
test_loader  = DataLoader(test_dataset,  batch_size=BATCH_SIZE,
                          shuffle=False, num_workers=2, pin_memory=True)

print(f"\nDataset → Train: {len(train_dataset)} | "
      f"Val: {len(val_dataset)} | Test: {len(test_dataset)}")


# ===============================================================
# 4. SURROGATE GRADIENT
# ===============================================================
class FastSigmoidSurrogate(torch.autograd.Function):
    @staticmethod
    def forward(ctx, x, gamma=1.0):
        ctx.save_for_backward(x)
        ctx.gamma = gamma
        return (x > 0).float()

    @staticmethod
    def backward(ctx, grad_out):
        (x,)  = ctx.saved_tensors
        g     = ctx.gamma
        return grad_out * (g / (1 + g * x.abs()) ** 2), None

spike_fn = FastSigmoidSurrogate.apply


# ===============================================================
# 5. LIF NEURON (Leaky Integrate-and-Fire)
# ===============================================================
class LIFNode(nn.Module):
    """
    Leaky Integrate-and-Fire neuron with surrogate-gradient spiking.

    Membrane update:
        v[t] = v[t-1] + (1/tau) * ( -(v[t-1] - v_rest) + I[t] )
    Spike:
        s[t] = 1 if v[t] > v_thresh else 0
    Reset:
        v[t] <- v_reset  if spiked (hard reset)
    """
    def __init__(self,
                 tau=2.0, v_rest=0.0,
                 v_reset=0.0, v_thresh=1.0,
                 gamma=1.0):
        super().__init__()
        self.tau      = tau
        self.v_rest   = v_rest
        self.v_reset  = v_reset
        self.v_thresh = v_thresh
        self.gamma    = gamma
        self.register_buffer('v', torch.tensor([]))

    def reset(self, device=None):
        dev = device or (self.v.device if self.v.numel() > 0 else DEVICE)
        if self.v.numel() > 0:
            self.v = torch.zeros_like(self.v).to(dev)

    def forward(self, I):
        if self.v.numel() != I.numel():
            self.v = torch.zeros_like(I)

        self.v = self.v.to(I.device)

        # Leaky integration
        dv     = (-(self.v - self.v_rest) + I) / self.tau
        self.v = self.v + dv

        # Spike generation (surrogate gradient)
        s = spike_fn(self.v - self.v_thresh, self.gamma)

        # Hard reset on spike
        self.v = self.v * (1 - s) + self.v_reset * s

        return s


# ===============================================================
# 6. CSNN_LIF_Deep — TRANSFER LEARNING MODEL
# ===============================================================
class CSNN_LIF_Deep(nn.Module):
    def __init__(self, time_window=TIME_WINDOW, freeze_backbones=True):
        super().__init__()
        self.time_window = time_window
        self.encoder     = encoding.PoissonEncoder()

        # ── Pretrained Backbones ────────────────────────────────────────
        mobilenet = models.mobilenet_v2(
            weights=models.MobileNet_V2_Weights.IMAGENET1K_V1)
        resnet    = models.resnet50(
            weights=models.ResNet50_Weights.IMAGENET1K_V1)

        self.mobilenet_features = mobilenet.features
        self.resnet_features    = nn.Sequential(*list(resnet.children())[:-1])

        if freeze_backbones:
            for p in self.mobilenet_features.parameters():
                p.requires_grad = False
            for p in self.resnet_features.parameters():
                p.requires_grad = False

        # ── BatchNorm on fused features (helps Poisson encoding) ─────────
        self.bn_input = nn.BatchNorm1d(1280 + 2048)   # 3328

        # ── Deep SNN Classifier (LIF neurons) ─────────────────────────────
        # Layer 1: 3328 → 512
        self.fc1   = nn.Linear(3328, 512)
        self.bn1   = nn.BatchNorm1d(512)
        self.lif1  = LIFNode()

        # Layer 2: 512 → 256
        self.fc2   = nn.Linear(512, 256)
        self.bn2   = nn.BatchNorm1d(256)
        self.lif2  = LIFNode()

        # Layer 3: 256 → 128
        self.fc3   = nn.Linear(256, 128)
        self.bn3   = nn.BatchNorm1d(128)
        self.lif3  = LIFNode()

        # Layer 4: 128 → 2 (output)
        self.fc4   = nn.Linear(128, 2)
        self.lif4  = LIFNode()

        # Spike rate tracking
        self.spike_rates = []

    def reset_states(self):
        for m in self.modules():
            if isinstance(m, LIFNode):
                m.reset(device=DEVICE)

    def forward(self, x):
        # ── STEP 1: CNN feature extraction — runs ONCE ─────────────────────
        with torch.no_grad():
            f_m = self.mobilenet_features(x)
            f_m = torch.mean(f_m, dim=[2, 3])     # GAP → (B, 1280)

            f_r = self.resnet_features(x)
            f_r = torch.flatten(f_r, 1)            # (B, 2048)

        fused = torch.cat((f_m, f_r), dim=1)       # (B, 3328)
        fused = self.bn_input(fused)               # normalise for encoder

        # ── STEP 2: SNN temporal loop — only FC layers repeat ──────────────
        spike_sum  = 0
        rate_accum = 0

        for t in range(self.time_window):

            # Poisson encoder re-samples every step → temporal diversity
            s = self.encoder(fused)                # (B, 3328) binary spikes

            # Layer 1
            y = self.fc1(s)
            y = self.bn1(y)
            y = self.lif1(y)
            rate_accum += y.mean().item()

            # Layer 2
            y = self.fc2(y)
            y = self.bn2(y)
            y = self.lif2(y)

            # Layer 3
            y = self.fc3(y)
            y = self.bn3(y)
            y = self.lif3(y)

            # Layer 4 — output
            y = self.fc4(y)
            y = self.lif4(y)

            spike_sum += y

        # Track mean spike rate of layer 1 (most representative)
        self.spike_rates.append(rate_accum / self.time_window)

        return spike_sum / self.time_window


# ===============================================================
# 7. ENERGY ESTIMATOR
# ===============================================================
def estimate_energy(model, time_window=TIME_WINDOW,
                    spike_rate=0.10, batch_size=1,
                    e_mac=4.6, e_ac=0.9, e_mem_sram=5.0):

    # CNN (runs ONCE)
    mobilenet_macs = 300e6
    resnet_macs    = 4100e6
    e_cnn          = (mobilenet_macs + resnet_macs) * e_mac * batch_size

    # SNN layers (run T times, scaled by spike rate)
    synapses = {
        'FC1 (3328→512)' : 3328 * 512,
        'FC2 (512→256)'  : 512  * 256,
        'FC3 (256→128)'  : 256  * 128,
        'FC4 (128→2)'    : 128  * 2,
    }
    e_snn_layers = {}
    for name, syn in synapses.items():
        e_snn_layers[name] = syn * spike_rate * e_ac * time_window * batch_size
    e_snn = sum(e_snn_layers.values())

    # Memory
    total_params = sum(p.numel() for p in model.parameters())
    e_mem        = total_params * e_mem_sram * batch_size

    total = e_cnn + e_snn + e_mem

    # Original model estimate (CNN runs T=10 times, ANN baseline)
    orig_cnn  = (mobilenet_macs + resnet_macs) * e_mac * 10 * batch_size
    orig_snn  = (3328*512 + 512*2) * spike_rate * e_ac * 10 * batch_size
    orig_mem  = (29e6) * e_mem_sram * batch_size
    orig_tot  = orig_cnn + orig_snn + orig_mem

    savings   = (orig_tot - total) / orig_tot * 100

    print("\n" + "="*60)
    print("  ENERGY ESTIMATE  (per inference, batch=1) — LIF MODEL")
    print("="*60)
    print(f"\n  CNN Backbones (runs ONCE):")
    print(f"    MobileNetV2:  {(mobilenet_macs*e_mac*batch_size)/1e9:.2f} GpJ")
    print(f"    ResNet-50:    {(resnet_macs   *e_mac*batch_size)/1e9:.2f} GpJ")
    print(f"    CNN total:    {e_cnn/1e9:.2f} GpJ")

    print(f"\n  SNN Layers (T={time_window}, spike_rate={spike_rate:.0%}):")
    for name, e in e_snn_layers.items():
        print(f"    {name}: {e/1e6:.4f} MpJ")
    print(f"    SNN total:    {e_snn/1e6:.4f} MpJ")

    print(f"\n  Memory (SRAM, {total_params/1e6:.1f}M params): {e_mem/1e6:.2f} MpJ")

    print(f"\n  {'─'*40}")
    print(f"  TOTAL (optimised):  {total/1e9:.4f} GpJ")
    print(f"  TOTAL (original):   {orig_tot/1e9:.2f} GpJ")
    print(f"\n  CNN share:   {e_cnn/total*100:.2f}%")
    print(f"  SNN share:   {e_snn/total*100:.4f}%")
    print(f"  Mem share:   {e_mem/total*100:.2f}%")
    print(f"\n  ✅ Energy saving vs original: {savings:.1f}%  "
          f"({orig_tot/total:.1f}× reduction)")
    print("="*60)
    return total


# ===============================================================
# 8. TRAINING
# ===============================================================
model     = CSNN_LIF_Deep(time_window=TIME_WINDOW).to(DEVICE)
criterion = nn.CrossEntropyLoss()
optimizer = torch.optim.Adam(
    filter(lambda p: p.requires_grad, model.parameters()), lr=LR)
scheduler = torch.optim.lr_scheduler.CosineAnnealingLR(
    optimizer, T_max=NUM_EPOCHS, eta_min=1e-6)

train_losses, train_accs = [], []
val_losses,   val_accs   = [], []

best_val_acc   = 0.0
best_weights   = None
patience_count = 0

# Print model summary
total_params     = sum(p.numel() for p in model.parameters())
trainable_params = sum(p.numel() for p in model.parameters() if p.requires_grad)
print(f"\nModel params  : {total_params/1e6:.2f}M total | "
      f"{trainable_params/1e6:.2f}M trainable")


def validate():
    model.eval()
    correct, total, loss_sum = 0, 0, 0.0
    with torch.no_grad():
        for data, target in val_loader:
            data, target = data.to(DEVICE), target.to(DEVICE)
            out      = model(data)
            loss_sum += criterion(out, target).item()
            preds    = out.argmax(dim=1)
            correct  += (preds == target).sum().item()
            total    += target.size(0)
            model.reset_states()
    return loss_sum / len(val_loader), 100 * correct / total


def train_epoch(epoch):
    model.train()
    model.spike_rates.clear()
    correct, total, loss_sum = 0, 0, 0.0

    for batch_idx, (data, target) in enumerate(train_loader):
        data, target = data.to(DEVICE), target.to(DEVICE)

        optimizer.zero_grad()
        out  = model(data)
        loss = criterion(out, target)

        # Spike rate regularisation
        if model.spike_rates:
            mean_rate     = sum(model.spike_rates) / len(model.spike_rates)
            spike_penalty = SPIKE_REG * (mean_rate - TARGET_RATE) ** 2
            loss          = loss + spike_penalty

        loss.backward()
        torch.nn.utils.clip_grad_norm_(model.parameters(), 5.0)
        optimizer.step()
        model.reset_states()

        loss_sum += loss.item()
        preds     = out.argmax(dim=1)
        correct  += (preds == target).sum().item()
        total    += target.size(0)

        if batch_idx % 20 == 0:
            rate_str = (f" | spike_rate={mean_rate:.3f}"
                        if model.spike_rates else "")
            print(f"  Epoch {epoch:02d} | Batch {batch_idx:04d} | "
                  f"Loss={loss.item():.4f}{rate_str}")

    return loss_sum / len(train_loader), 100 * correct / total


print("\n" + "="*60)
print("  TRAINING — CSNN_LIF_Deep")
print("="*60)

for epoch in range(1, NUM_EPOCHS + 1):
    t0              = time.time()
    tr_loss, tr_acc = train_epoch(epoch)
    v_loss,  v_acc  = validate()
    scheduler.step()

    train_losses.append(tr_loss); train_accs.append(tr_acc)
    val_losses.append(v_loss);    val_accs.append(v_acc)

    elapsed = time.time() - t0
    print(f"\nEpoch {epoch:02d}/{NUM_EPOCHS}  ({elapsed:.1f}s) → "
          f"Train Loss={tr_loss:.4f}  Acc={tr_acc:.2f}% | "
          f"Val Loss={v_loss:.4f}  Acc={v_acc:.2f}%\n")

    # Early stopping
    if v_acc > best_val_acc:
        best_val_acc   = v_acc
        best_weights   = copy.deepcopy(model.state_dict())
        patience_count = 0
        print(f"  ✓ New best val acc: {best_val_acc:.2f}% — weights saved")
    else:
        patience_count += 1
        if patience_count >= PATIENCE:
            print(f"  Early stopping at epoch {epoch} "
                  f"(no improvement for {PATIENCE} epochs)")
            break

model.load_state_dict(best_weights)
print(f"\nTraining complete.  Best Val Acc: {best_val_acc:.2f}%")


# ===============================================================
# 9. TRAINING CURVES
# ===============================================================
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

axes[0].plot(train_losses, marker='o', label='Train Loss')
axes[0].plot(val_losses,   marker='o', label='Val Loss')
axes[0].set_title('Loss vs Epoch'); axes[0].set_xlabel('Epoch')
axes[0].set_ylabel('Loss'); axes[0].grid(True); axes[0].legend()

axes[1].plot(train_accs, marker='o', label='Train Acc')
axes[1].plot(val_accs,   marker='o', label='Val Acc')
axes[1].set_title('Accuracy vs Epoch'); axes[1].set_xlabel('Epoch')
axes[1].set_ylabel('Accuracy (%)'); axes[1].grid(True); axes[1].legend()

plt.tight_layout()
plt.show()


# ===============================================================
# 10. TEST EVALUATION
# ===============================================================
def evaluate(loader):
    model.eval()
    y_true, y_pred = [], []
    with torch.no_grad():
        for data, labels in loader:
            data, labels = data.to(DEVICE), labels.to(DEVICE)
            out   = model(data)
            preds = out.argmax(dim=1)
            y_true.extend(labels.cpu().numpy())
            y_pred.extend(preds.cpu().numpy())
            model.reset_states()
    return y_true, y_pred

y_true, y_pred = evaluate(test_loader)

print('\nFinal Test Accuracy:', accuracy_score(y_true, y_pred))
print('\nClassification Report:')
print(classification_report(y_true, y_pred,
                             target_names=['Not Fire', 'Fire']))

cm = confusion_matrix(y_true, y_pred)
plt.figure(figsize=(6, 5))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues',
            xticklabels=['Not Fire', 'Fire'],
            yticklabels=['Not Fire', 'Fire'])
plt.title('Confusion Matrix — CSNN_LIF_Deep')
plt.tight_layout()
plt.show()




# ===============================================================
# 11. ENERGY ESTIMATION
# ===============================================================
obs_rate = (sum(model.spike_rates) / len(model.spike_rates)
            if model.spike_rates else 0.10)
print(f'\nObserved mean spike rate: {obs_rate:.3f}')

estimate_energy(model, time_window=TIME_WINDOW,
                spike_rate=obs_rate, batch_size=1)


# ===============================================================
# 12. SAVE MODEL
# ===============================================================
save_path = '/content/drive/MyDrive/csnn_lif_deep_final.pth'
torch.save({
    'model_state_dict' : model.state_dict(),
    'optimizer_state_dict': optimizer.state_dict(),
    'best_val_acc'     : best_val_acc,
    'config': {
        'time_window'  : TIME_WINDOW,
        'spike_reg'    : SPIKE_REG,
        'target_rate'  : TARGET_RATE,
        'architecture' : '3328→512→256→128→2 (LIF)',
    }
}, save_path)
print(f'\nModel saved → {save_path}')
```
